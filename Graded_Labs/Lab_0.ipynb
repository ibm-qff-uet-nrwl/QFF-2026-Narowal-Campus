## لیب 1: ایک کیوبٹ کی اسٹیٹ (Single Qubit State)

**مقصد:** ایک کوانٹم سرکٹ بنائیں جس میں صرف 1 کیوبٹ ہو اور وہ |+> اسٹیٹ میں ہو۔

**ہدایات:**
1. `QuantumCircuit(1)` سے ایک کیوبٹ کا سرکٹ بنائیں۔
2. اس کیوبٹ پر ہڈامارڈ (Hadamard) گیٹ لگائیں۔

**نوٹ:** نیچے دیے گئے کوڈ میں، جہاں `#` لکھا ہے وہ کمنٹ ہے اور یہ کوڈ نہیں چلتا۔ لیب حل کرنے کے لیے آپ کو `#` ہٹانا ہوگا۔

</div>

```python
from qiskit import QuantumCircuit
from grader import grade_lab1_single_qubit

# لیب 1: |+> اسٹیٹ بنائیں
qc1 = QuantumCircuit(1)

# یہاں اپنا کوڈ لکھیں:
qc1.h(0)  # <-- لیب حل کرنے کے لیے اس لائن کے شروع سے # ہٹا دیں

# اپنے سرکٹ کو گریڈر کے پاس بھیجیں
result = grade_lab1_single_qubit(qc1)

print("="*50)
print(f"Score: {result['score']}/{result['max']}")
print(result['msg'])
print("="*50)
```
