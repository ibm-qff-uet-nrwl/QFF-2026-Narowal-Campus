## Lab 2: Multi-Qubit Gates (CNOT)

**Objective:** Learn how to use the CNOT gate to make two qubits interact.

**The CNOT Gate:**
The CNOT (Controlled-NOT) gate acts on two qubits:
- **Control Qubit:** Decides whether to act.
- **Target Qubit:** Gets flipped if the control is |1>.

**Real-life example (The Security Guard):**
Think of the control qubit as a security guard and the target as a door. The door only opens (flips) if the guard says "yes" (control is 1).

**Instructions:**
1. Initialize a `QuantumCircuit` with 2 qubits.
2. Apply a Hadamard gate to qubit 0.
3. Apply a CNOT gate with control=0 and target=1.

## لیب 2: ملٹی کیوبٹ گیٹس (CNOT)

**مقصد:** CNOT گیٹ کا استعمال سیکھیں تاکہ دو کیوبٹس آپس میں تعامل کر سکیں۔

**CNOT گیٹ:**
CNOT (کنٹرولڈ-ناٹ) گیٹ دو کیوبٹس پر کام کرتا ہے:
- **کنٹرول کیوبٹ:** فیصلہ کرتا ہے کہ عمل کرنا ہے یا نہیں۔
- **ٹارگٹ کیوبٹ:** اگر کنٹرول |1> ہو تو یہ پلٹ جاتا ہے۔

**حقیقی زندگی کی مثال (سیکیورٹی گارڈ):**
کنٹرول کیوبٹ کو ایک سیکیورٹی گارڈ سمجھیں اور ٹارگٹ کو ایک دروازہ۔ دروازہ صرف اسی وقت کھلتا (پلٹتا) ہے جب گارڈ "ہاں" کہے (کنٹرول 1 ہو)۔

**ہدایات:**
