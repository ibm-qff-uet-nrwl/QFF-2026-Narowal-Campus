## Qubits

### States as Vectors

***Qubits*** are often represented with a column vector called a ***ket***.  

Each ket has an associated row vector called a ***bra***.

$$
\begin{bmatrix}\phantom{A}\!\!\! \alpha \\ \beta\end{bmatrix}
\qquad \text{and} \qquad
\begin{bmatrix} \,\alpha^* & \beta^* \, \end{bmatrix}
$$

$\alpha$ and $\beta$ are complex numbers, and $\alpha^*$ and $\beta^*$ are their ***complex conjugates***.

Complex conjugation means that if $z = x + iy$, then $z^* = x - iy$.

Importantly, the state represented by $\alpha$ and $\beta$ is normalized:
$$
\vert \alpha \vert^2 + \vert \beta \vert ^2 = 1
$$

> Notes on terminology:  
> The term for converting a ket to a bra, or a bra to a ket, is taking the ***conjugate transpose***.    
> A quantum state follows the ***Euclidean*** norm.

### Vectors Need a Basis

We always choose a basis to work in.  The ***standard*** or ***computational basis*** states
in quantum computing are:
$$
\vert 0 \rangle = \begin{bmatrix} 1 \\ 0 \end{bmatrix}   \qquad
\text{and}
\qquad \vert 1 \rangle = \begin{bmatrix} 0 \\ 1 \end{bmatrix}
$$

A qubit in a ***pure state*** can be written in several equivalent ways:

$$
\vert \psi \rangle
\ \ =\ \
\alpha \vert 0 \!\: \rangle + \beta \vert 1\!\: \rangle
\ \ =\ \
\alpha \begin{bmatrix}1\\0\end{bmatrix} + \beta \begin{bmatrix}0\\1\end{bmatrix}
\ \ = \ \
\left[\begin{matrix}
\phantom{A}\!\!\!\alpha \\ \beta
\end{matrix}\right]
$$

This linear combination of basis states is called superposition.

> If a state can be described by a single state vector it is a pure state.
>
> The counterpart to pure states is the ***mixed state***, which will be introduced with noise on day five.


## Measurement

### Coefficients are Probability Amplitudes

By the ***Born rule***, probabilities are the squared magnitudes of the amplitudes.

The ***probability amplitudes*** for $\vert 0 \rangle$ and $\vert 1 \rangle$ are the coefficients $\alpha$ and $\beta$, so:
$$
P(0) = \vert \alpha \vert ^2 \qquad \text{and} \qquad P(1) = \vert \beta \vert^2
$$

This is why the coefficients must be normalized.
$$
\lvert \alpha \rvert ^2 + \lvert \beta \rvert^2 = 1
$$

Note, if $\alpha = x + iy$ then:
$$
\vert \alpha \vert ^2
\ = \
\alpha^* \alpha
\ = \
(x-iy)(x+iy)
\ =\
x^2 + y^2
$$

> The probabilities of all possible outcomes must sum to one.   
> This is the why quantum states are required to be normalized.


```python
from qiskit.quantum_info import Statevector

psi = Statevector([0.6, 0.8])
p = psi.probabilities_dict()

print('P(0) =', p['0'])
print('P(1) =', p['1'])
print()
print('P(0) + P(1) =', p['0'] + p['1'])
```

### Finding an Outcome

Measuring in the computational basis gives either 0 or 1.
$$
\vert \psi \rangle
= \frac{\sqrt 3}{2} \vert 0 \rangle +
\frac1 2 \vert 1 \rangle
$$

Applying the Born rule to $\vert \psi \rangle$, the probabilities of measuring either 0 or 1 are:
$$
P(0) = \left \vert \frac{\sqrt 3} 2 \right \vert ^2 = \ \frac 3 4
\qquad \qquad
P(1) = \left\vert \frac{1}{2} \right\vert^2 = \  \frac 1 4
$$

After measurement, the qubit's state becomes the basis state of the measured outcome.
$$
0 \rightarrow \vert 0\rangle \qquad \qquad 1 \rightarrow \vert 1 \rangle
$$

> For any single measurement, only one outcome will be produced.  Which outcome occurs is random, based on the probabilities given by Born rule.
>
> The measured outcome becomes the new state.  If immediately measured again in the same basis, the same outcome will be obtained.
>
> This change from an arbitrary superposition to a single basis state is often referred to as ***collapse***.

```python
for k, v in psi.sample_counts(1000).items():
    print(v, 'of 1000 measurements had outcome', k)
```

## Multiple Qubits
### Two Qubits, Four Basis States

Two qubits have four associated basis states:
$$
\vert 00 \rangle, \  \vert 01 \rangle, \  \vert 10\rangle, \  \vert 11 \rangle
$$

The associated state vector has four terms:
$$
\vert \psi \rangle
=
a\vert 00 \rangle
+
b \vert 01 \rangle
+
c \vert 10 \rangle
+
d \vert 11 \rangle
$$

Amplitudes determine the probabilities of each result, so they must be normalized:
$$\vert a \vert ^2
+
\vert b \vert ^2
+
\vert c \vert ^2
+
\vert d \vert ^2
=
1
$$

> Generally, with $N$ qubits there are $2^N$ computational basis states, each with its own amplitude.



### Combining States

The ***tensor product*** (also known as the ***Kronecker product***) combines the states of multiple qubits.
$$
\vert 0 \rangle
\otimes
\vert 1 \rangle
=
\vert 01 \rangle
\qquad \qquad
\vert 1 \rangle
\otimes
\vert 0 \rangle
=
\vert 10 \rangle
$$


Generally:
$$
\begin{bmatrix}
a \\ b
\end{bmatrix}
\otimes
\begin{bmatrix}
c \\ d
\end{bmatrix}
=
\begin{bmatrix}
ac \\
ad \\
bc \\
bd
\end{bmatrix}
$$

> To save space in kets the tensor product symbol is almost always dropped: $\vert 0 \rangle \otimes \vert 1 \rangle = \vert 01 \rangle$
>
> Each additional qubit doubles the size of the resulting state vector.

```python
sv0 = Statevector.from_label('0')
sv1 = Statevector.from_label('1')

sv01 = sv0.tensor(sv1)

print(sv01 == Statevector.from_label('01'))
```

### States We Can Separate

If a multi-qubit state can be written as a tensor product of individual pure states it is called a ***product state***.
$$
\vert \!\:{+}\!\: \rangle = \frac{\vert 0\rangle+\vert 1 \rangle}{\sqrt 2}
$$
$$
\vert \!\:{+}\!\: \rangle
\otimes
\vert \!\:{+}\!\: \rangle
= \frac{1}{2} \Big(
\vert 00 \rangle +
\vert 01 \rangle +
\vert 10 \rangle +
\vert 11 \rangle
\Big)
$$

However, some states cannot be written as a product state, for example:
$$\vert \Phi^+ \rangle = \frac{\vert 00 \rangle + \vert 11 \rangle}{\sqrt 2}$$

```python
plus = Statevector.from_label('+')

plus.tensor(plus).draw('latex')
```

### States We Cannot Separate

When a state cannot be written as a tensor product of individual pure states it is
called an ***entangled state***.
$$
\vert \Phi^+ \rangle = \frac{\vert 00 \rangle + \vert 11 \rangle}{\sqrt 2}
$$

A measurement of either qubit determines the outcome of a measurement of the other qubit.

For $\vert \Phi^+ \rangle$, the only possible outcomes for a measurement in the computational
basis are $\vert 00 \rangle$ and $\vert 11 \rangle$, each with probability $1/2$.

> This is one of four ***Bell states**, a standard set of maximally entangled two-qubit states.


```python
from math import sqrt

bell = Statevector([1/sqrt(2), 0, 0, 1/sqrt(2)])

for k, v in bell.probabilities_dict().items():
    print(f'P({k}) = {v}')
```

## Operators

### Matrices Act on States

***Operators*** are linear transformations that act on the vectors representing quantum states.  They are represented by matrices:
$$
X = \begin{bmatrix}
0 & 1 \\
1 & 0
\end{bmatrix}
$$

If an operator $A$ acts on $\vert \psi \rangle$, then:  $\ \ A \vert \psi \rangle = \vert \psi' \rangle$
$$
X \vert 0 \rangle =
\begin{bmatrix}
0 & 1 \\
1 & 0
\end{bmatrix}
\begin{bmatrix}
1 \\ 0
\end{bmatrix} =
\begin{bmatrix}
0 \\ 1
\end{bmatrix}
$$

$$
X \vert 1 \rangle =
\begin{bmatrix}
0 & 1 \\
1 & 0
\end{bmatrix}
\begin{bmatrix}
0 \\ 1
\end{bmatrix} =
\begin{bmatrix}
1 \\ 0
\end{bmatrix}
$$

> This operator $X$ is known as the ***bit-flip*** operator.
>
> Not every matrix represents a physically allowed operation.  We will talk about the required properties soon.


```python
from qiskit.quantum_info import Operator

X = Operator.from_label('X')
I = X.dot(X)

sv0 = Statevector.from_label('0')
sv1 = Statevector.from_label('1')

print('X|0> = |1> ? ', sv0.evolve(X) == sv1)
print('X|1> = |0> ? ', sv1.evolve(X) == sv0)
```

### Evolution and Measurement

**Unitary operators** describe reversible evolutions of quantum states.  Matrices are unitary when:
$$
U^\dagger U = UU^\dagger = I
$$

***Hermitian operators*** describe ***observables***, their real eigenvalues are the possible measurement outcomes for that observable.
$$
A^\dagger = A
$$


> The symbol $\dagger$, read as “dagger,” indicates the conjugate transpose operation, also known as the adjoint.
>
> Multiplication by a unitary matrix preserves normalization when acting on a state vector.  Later you will see how this preserves probabilities.
>
> Hermitian operators are their own conjugate transpose. Because of this, their eigenvalues are real numbers.  Since those are the possible measurement outcomes of the operator, they must be real valued.

## Noise in Quantum Computers

### Where Does Noise Come From?

Real qubits are physical systems.

Qubits interact with their environment, the QPU's control operations are
imperfect, and the final measurements can be incorrect.

These are all sources of noise:
- Decoherence accumulates as time passes.
- Gate errors are caused by imperfections in the operations applied.
- Readout errors happen when the measured result is incorrect.

> The rate of different errors can vary widely even between qubits on the same QPU.
>
> Decoherence is continuous and just happens as time passes, but gate and readout errors are caused by operations on the qubits.
>
> Measurement relies on amplifying very small signals and can easily give the wrong result.

## Describing Noise

### State Vectors Are Not Enough


Until now we have represented pure states as state vectors.  For a single qubit:
$$
\vert \psi \rangle = \alpha \vert 0 \rangle + \beta\vert 1 \rangle =
\left [ \begin{matrix}
\phantom{A}\!\!\!\alpha \\ \beta
\end{matrix} \right ]
$$

States can also be more generally represented as density matrices:
$$
\rho =
\vert \psi \rangle \langle \psi \vert
= \left [ \begin{matrix}
\phantom{I}\!\!\alpha \\ \beta
\end{matrix} \right ]
\begin{bmatrix}
\alpha^* & \beta^*
\end{bmatrix}
=
\left [
\begin{matrix}
\phantom{A} \!\!
\alpha\alpha^* & \alpha\beta^* \\
\beta\alpha^*  & \beta\beta^*
\end{matrix}
\right]
$$

> The operation shown above is the ***outer product***.
>
> Density matrices represent more than just pure states.



```python
from qiskit.quantum_info import DensityMatrix, Statevector

psi = Statevector([0.6, 0.8])
rho = DensityMatrix(psi)

print(rho)
```

### Interpreting a Density Matrix

For a single qubit:
$$
\rho
=
\left [
\begin{matrix}
\phantom{A} \!\!
\vert\alpha\vert^2 & \alpha\beta^* \\
\beta\alpha^*  & \vert\beta\vert^2
\end{matrix}
\right]
$$

The ***diagonal elements*** are the probabilities $P(0)$ and $P(1)$.

The ***off-diagonal elements*** describe the ***coherence*** between the basis states.

Valid density matrices are positive semi-definite and have the properties:
$$
\operatorname{Tr}(\rho) = 1  
\qquad\qquad
\rho = \rho^\dagger
$$

> The diagonal entries are sometimes called "populations" or "population information".
>
> Off-diagonal entries contain both magnitude of coherence and its relative phase.
>
> $Tr(\rho)$ is read as "the trace of rho" and is the sum of the diagonal entries.  
> This makes sense because probabilities must sum to 1.
>
> Positive semi-definite meand the matrix is both Hermitian and all of its eigen values are non-negative.  
> This ***is*** important, but not for the limited use of density matrices in this course.
>
> See John Watrous' course on the [General Formulation of Quantum Information](https://quantum.cloud.ibm.com/learning/en/courses/general-formulation-of-quantum-information).

```python
import numpy as np

print('Tr(rho) = 1?      ', rho.trace() == 1)
print('rho is Hermitian? ', np.allclose(rho.data, rho.data.conj().T))

# This is a bit of a cheat since DensityMatrix has no .is_hermitian() or .adjoint().
# You could also use:  rho.to_operator().adjoint().equiv(rho)

```

### Why Density Matrices Matter

Weighted combinations of density matrices are also density matrices when the
weights are probabilities.
$$
\rho = p_1\rho_1 + p_2\rho_2+\dots
$$

The state is $\rho_1$ with probability $p_1$, it is state $\rho_1$ with probability $p_1$, etc...

Probabilities still need to be non-negative and add up to one.
$$
p_i \geq 0, \qquad \sum_i p_i =1
$$

> The technical term for this weighted combination is a ***convex combination***.




```python
rho_0 = DensityMatrix(Statevector.from_label('0'))
rho_1 = DensityMatrix(Statevector.from_label('1'))

rho_mixed = 0.5 * rho_0 + 0.5 * rho_1

print(rho_mixed)
```

### Mixed States


What if our state had a 50% chance of being $\vert 0 \rangle$ or $\vert 1 \rangle$?

By measuring in the computational basis can we differentiate it from $\vert + \rangle$?
$$
\vert \!+\!\,\rangle \ = \
\frac{|0\rangle + |1\rangle}{\sqrt{2}},
\qquad\qquad
\rho_+ \ = \
\begin{bmatrix}
1/2 & 1/2\\
1/2 & 1/2
\end{bmatrix}
$$

These states cannot be distinguished with only a $Z$-basis measurement.

> $\rho$ here is a mixed state that cannot be expresed as a simple state vector.
>
> The states have the same probabilities, but different coherences.
>
> An $X$-basis measurement *CAN* distinguish the states.  This shows that the off-diagonal entries do contain physically meaningful information.

```python
rho_plus = DensityMatrix(Statevector.from_label('+'))

print('rho_mixed probabilities: ', rho_mixed.probabilities())
print('rho_plus probabilities:  ', rho_plus.probabilities())
print()
print('rho_mixed = rho_plus ?   ', rho_mixed == rho_plus)
```

### Operators on Density Matrices


Unitary operators evolving state vectors is represented by:
$$ U \vert \psi \rangle  =   \vert \psi'\!\:\rangle $$

If $\vert \psi'\!\:\rangle = U \vert \psi \rangle$, then $\langle \psi'\vert = \langle \psi \vert U^\dagger$ and we can write the expression:
$$
\vert \psi' \!\: \rangle \langle \psi'\vert
\ =\
U \vert \psi \rangle \langle \psi \vert U^\dagger
$$

So the same unitary operator evolving a density matrix is written as:
$$
U \rho U^\dagger  =   \rho'
$$

> The order in $\langle \psi'\vert = \langle \psi \vert U^\dagger$ is reversed by the adjoint product rule: $(AB)^* = B^*A^*$
>
> In general, a process that maps density matrices to density matrices is called a quantum channel.  
> Quantum channels are beyond the scope of this course, see the Watrous course.

```python
# Remember that @ is the symbol for matrix multiplication.
U_rho_Udag = DensityMatrix(X @ rho_0 @ X.adjoint())

# DensityMatrix provides the .evolve() method.
print('Same result? ', U_rho_Udag == rho_0.evolve(X))
```

## Noise Models

### Representing an Imperfect Operation

A ***noise model*** gives a mathematical rule for approximating how a type of noise
changes the state.

The model is **NOT** a description of the exact physical process causing the noise.

The model **IS** a mathematical tool that replicates the statistics of the physical
process causing the noise.



### Randomizing the State

The ***maximally mixed state*** is the state of randomness represented by:
$$
\frac I 2
\  =\  
\frac 1 2 \!\begin{bmatrix} 1 & 0 \\ 0 & 1\end{bmatrix}
\ =\  
\begin{bmatrix} 1/2 & 0 \\ 0&1/2\end{bmatrix}
$$

***Depolarizing noise*** replaces some fraction of the state with the completely mixed state.
For a single qubit this can be represented as:
$$
\rho \rightarrow (1-\lambda)\rho +\lambda\frac I 2
$$

The term $\lambda$ is in the range $0 \le \lambda \le 1$, where $\lambda = 0$ represents no noise and $\lambda = 1$ will completely depolarize $\rho$.

> As a state depolarizes its Bloch vector becomes shorter.  
> Pure states exist on the surface of the Bloch sphere.  
> The maximally mixed state is in the center of the Bloch sphere.
>
> Depolarization affects both the popuation information (on-diagonals) and the coherence information (of-diagonals).
>
> Diagonal entries move towards $1/2$, while off-diagonal entries move towards $0$.

### Depolarization in Action

A qubit in the state $\vert 0 \rangle$ undergoes depolarization.  Remember that:
$$
\vert 0 \rangle \langle 0 \vert \ =\
\begin{bmatrix}
1 & 0\\
0 & 0
\end{bmatrix}
\qquad \text{and} \qquad
\frac I 2 =
\begin{bmatrix}
1/2 & 0 \\
0 & 1/2
\end{bmatrix}
$$

So, if $\lambda = 1/4$:

$$
\rho' = \frac 3 4 \rho + \frac 1 4 \frac I 2
\ \ =\ \
\frac 3 4 \!
\begin{bmatrix}
1 & 0 \\
0 & 0
\end{bmatrix}
+
\frac 1 4 \!
\begin{bmatrix}
1/2 & 0 \\
0 & 1/2
\end{bmatrix}
\ = \
\begin{bmatrix}
7/8 & 0 \\
0 & 1/8
\end{bmatrix}
$$



### Random Gates

Pauli noise models the result of errors as $X$, $Y$, and $Z$ operations which occur with specified probabilities.

Pauli noise can model random ***bit flips*** ($X$), ***phase flips*** ($Z$), or a combined bit-
and-phase flip ($Y$) occurring during computation.

$$
\rho \rightarrow p_I\rho +
p_XX\rho X^\dagger +
p_YY\rho Y^\dagger +
p_ZZ\rho Z^\dagger
$$

The probabilities must still be non-negative and their sum must be one.

> Notice that each term is just a normal unitary evoltion of the density matrix.
>
> $p_I$ is the probability that nothing happens.
>
> The final noisy state is just a mixture of the four elements weighted by their probabilities.


### Pauli Noise in Action

Starting with $\rho = \vert 0 \rangle\langle 0 \vert$ and supposing all four probabilities are $1/4$.

$$
\rho = Z\rho Z^\dagger =  
\begin{bmatrix}1 & 0 \\ 0 & 0 \end{bmatrix},
\qquad
X\rho X^\dagger = Y\rho Y^\dagger = \begin{bmatrix}0 & 0 \\ 0 & 1 \end{bmatrix}
$$

This gives:
$$
\rho'
\ =\
\frac 1 2 \!
\begin{bmatrix}
1 & 0 \\
0 & 0
\end{bmatrix}
+
\frac 1 4 \!
\begin{bmatrix}
0 & 0 \\
0 & 1
\end{bmatrix}
+
\frac 1 4 \!
\begin{bmatrix}
0 & 0 \\
0 & 1
\end{bmatrix}
\ =\  
\begin{bmatrix}
1/2 & 0 \\
0 & 1/2
\end{bmatrix}
$$


Equal probability Pauli noise has completely randomized the state.

> Note that $Z$ has no *OBSERVABLE* effect on $\vert0\rangle$ and $\vert 1 \rangle$.  
> In the computational basis $Z$ errors only affect relative phase.
>
> It might seem odd that $X\rho X$ and $Y \rho Y$ are the same matrix, but you multiply the $1$ by $i$ and $-i$.
>
> Global phase factors are cancelled by the outer product that generates the density matrix.


### Relaxation and Decoherence

Two parameters are used to describe time dependent noise: $T_1$ and $T_2$

$T_1$ is the energy relaxation time.  It describes the time for an excited qubit to decay from $\vert0\rangle$ to $\vert 1\ rangle$.

$T_2$ is the phase coherence time.  It describes how quickly coherence is lost between $\vert 0 \rangle$  and $\vert 1 \rangle$.

Each parameter affects different parts of the density matrix.

> Commonly are called the "relaxation time" and "dephasing time".
>
> $T_1$ lowers the probability of non-zero states on the diagonal entries.
> $T_2$ lowers the magnitude of the off-diagonal coherence values.
>
> Both $T_1$ and $T_2$ are real times measured fromt he behavior of the physical qubits.


### Energy Relaxation

For a single qubit density matrix:
$$
\rho
=
\begin{bmatrix}
\rho_{00} & \rho_{01} \\
\rho_{10} & \rho_{11}
\end{bmatrix}
$$

The probability of finding the qubit in an excited-state, given by $\rho_11$, decreases as:

$$
\rho_{11}(t) = e^{-t/T_1} \rho_{11}
$$

The lost probability moves to the ground state, $\rho_00$:
$$
\rho_{00}(t) =
\rho_{00} +
(1 - e^{-t/T_1}) \rho_{11}
$$

Larger $T_2$ values mean the excited-state probability decays more slowly.

> $e^{-x}$ is the same as $1/e^x$.  
> This starts at $1$ for $X=0$ and approaches $0$ as $x$ grows towards infinity.


### Phase Coherence

For a single qubit density matrix:
$$
\rho
=
\begin{bmatrix}
\rho_{00} & \rho_{01} \\
\rho_{10} & \rho_{11}
\end{bmatrix}
$$

the coherence between $\vert 0 \rangle$ and $\vert 1 \rangle$, given by the off-diagonal terms, decreases
as:
$$
\rho_{01}(t) = e^{-t/T_2} \rho_{01}
\qquad\qquad
\rho_{10}(t) = e^{-t/T_2} \rho_{10}
$$

Larger $T_2$ values mean coherence is retained for longer.

> $T_2$ affect off-diagonal terms, diagonal probabilities are unchanged.


### Thermal Relaxation

Putting these effects together gives us a model of how the density matrix
changes over time:

$$
\rho(t) =
\begin{bmatrix}
\rho_{00} + (1 - e^{-t/T_1}) \rho_{11} & \ \  &
e^{-t/T_2} \rho_{01} \\
e^{-t/T_2} \rho_{10} & &
e^{-t/T_1} \rho_{11} \\
\end{bmatrix}
$$


### Measuring and $T1$ and $T2$

The values of $T_1$ and $T_2$ are derived experimentally for each qubit:

$$
\text{Prepare }\vert 1 \rangle\  \rightarrow\ \text{Wait}\ \rightarrow\ \text{Measure in }Z\text{-basis}\quad(T_1)
$$
$$
\text{Prepare }\vert + \rangle\  \rightarrow\ \text{Wait}\ \rightarrow\ \text{Measure in }X\text{-basis}\quad(T_2)
$$

This experiment is done for many shots at different wait durations, and the
resulting statistics are used to determine the $T_1$ and $T_2$ times for each qubit.

> Values vary from qubit to qubit.
>
> Results from each duration give a point on the decay curve.
>
> Once data is gathered an exponential decay curve is fit.
>
> "Measure in X-basis" means to apply a Hadamard gate before measuring normally.

## Simulating Noise

### Ideal vs. Noisy Simulation

An ideal simulation assumes perfect qubits and gates.

***Qiskit Aer*** is a high-performance simulator that supports noise models.

Two common ways to simulate noise in Aer are to
- use a ***custom noise model*** or to
- use ***calibration data from a real backend***.



```python
from qiskit_aer import AerSimulator
from qiskit_aer.noise import NoiseModel
from qiskit_ibm_runtime.fake_provider import FakeFez

ideal_sim   = AerSimulator()
noisy_sim   = AerSimulator(noise_model=NoiseModel())
backend_sim = AerSimulator.from_backend(FakeFez())
```

### Simulating Noise Simulating From a Backend

Aer can use the properties of a real or fake backend.

The backend information includes properties like the QPU's basis gates, qubit
coupling map, and noise properties derived from device calibration data.

This can be used to preview how a circuit should behave on real hardware.

Using fake backends we can work locally, without an internet connection.

> Fake backends provide previous calibration data without needing to connect to IBM.
>
> Real backends provide current calibration data, but require connecting to the IBM runtime.
>
> The goal is to replicate the hardware as closely as possible, but it still is just a simulator.


```python
# Build a test circuit
n = 8

qc = QuantumCircuit(n)

qc.h(0)
for i in range(1, n):
    qc.cx(0, i)

qc.measure_all()

qc.draw('mpl')
```

```python
# Transpile the circuit
from qiskit.transpiler import generate_preset_pass_manager
pm = generate_preset_pass_manager(AerSimulator())

qc_isa = pm.run(qc)
```

```python
# Ideal simulation baseline
ideal = ideal_sim.run(qc_isa).result()
ideal.get_counts()
```

```python
# Visualize the ideal output
from qiskit.visualization import plot_histogram

plot_histogram(ideal.get_counts())
```

```python
# Use the simulator configured from a real backend
qc_isa = generate_preset_pass_manager(FakeFez()).run(qc)
backend_res = backend_sim.run(qc_isa).result()
```

```python
# Visualize the results
plot_histogram(backend_res.get_counts())
```

### Building a Custom Noise Model

The `qiskit_aer.noise` package provides `NoiseModel` and functions to define each of the noise models we have already discussed:

- `depolarizing_error(param, num_qubits)`
- `pauli_error(noise_ops)`
- `thermal_relaxation_error(t1, t2, time, excited_state_population=0)`

> For `depolarizing_error`, `param` is $\lambda$.
>
> `pauli_error` expects `noise_ops` to be a list of tuples.
>
> The `t1`, `t2`, `time` parameters to `thermal_relaxation_error` are expected to be in nanoseconds.

```python
from qiskit_aer.noise import depolarizing_error, pauli_error, thermal_relaxation_error

error1 = depolarizing_error(0.03, 2)
error2 = pauli_error([('I', 0.95),('X', 0.03),('Y',0.01),('Z',0.01)])
error3 = thermal_relaxation_error(60_000, 40_000, 100_000)
```

```python
noise_model = NoiseModel()
noise_model.add_all_qubit_quantum_error(error1, ['cx'])
```

```python
noisy_sim = AerSimulator(noise_model=noise_model)

pm = generate_preset_pass_manager(noisy_sim)

noisy_res = noisy_sim.run(pm.run(qc)).result()
noisy_res.get_counts()
```
