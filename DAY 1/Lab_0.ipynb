## Quantum Computing

### What Does Quantum Mean?

**Quantum mechanics** is the theory used to describe the behavior of physical systems
at very small scales.

**Quantum computing** uses these behaviors to perform calculations.

We will not be covering all of quantum mechanics, but there are some important
concepts that we will use:
- Physical systems have a state.
- Operations can change, or evolve, the state.
- Measurements of the system are probabilistic in nature.

> Note:  
> The **state** of a system is everything we need to know to predict that system's future behavior.  
> Quantum measurements are probabilistic, but some outcomes can still have a 100% probability.  


### The Qubit

Classical computers use ***bits*** when they calculate.

Quantum computers use quantum bits, called ***qubits***, for their calculations.

A qubit is a physical system that can be controlled as a ***two-state quantum system***.

Qubits can be made from many kinds of quantum systems:
- electron or nuclear spin,
- energy levels in an atom or ion,
- photon polarization, or
- energy levels in a superconducting circuit.

> The system can (and most of the time will) have more than two states.
>  
> The key is being able to isolate, control, and distinguish those two states.


### States and Evolution

A quantum computation can be viewed as:

$$
\text{Initial State} \rightarrow
\text{Operations} \rightarrow
\text{Final State}
$$

Our goal is to use our knowledge of the rules of quantum state evolution to
perform mathematical calculations that would otherwise require far too many
resources to classically implement.

The way these quantum rules will be described is with **vectors** and **matrices**.

> Consider how a problem might be given in a quantum mechanics classroom:  
>
> The students
> are given information about the arrangement of some quantum objects and their local
> environment.  Then they are asked to calculate some future state of the system.
>  
> If we are clever about arranging quantum objects and the environment they are in,
> then we can let the system evolve and extract the result of a calculation by
> measuring the system.



## Vectors and Matrices

### Vectors


Vectors are 1-dimensional arrays of numbers.

They can be ***row vectors*** or ***column vectors***:
$$
\begin{bmatrix}
1 & 2 & 3 &4
\end{bmatrix}
\qquad
\begin{bmatrix}
3 \\ 6 \\ 9
\end{bmatrix}
$$


Vector addition:
$$
\begin{bmatrix}
\phantom{1}\!\!\! a \\ \phantom{1}\!\!\!b
\end{bmatrix}
+ \begin{bmatrix}
\phantom{1}\!\!\!c \\ \phantom{1}\!\!\!d
\end{bmatrix}
= \begin{bmatrix}
\phantom{1}\!\!\!a \!\,+\!\, c \\ \phantom{1}\!\!\!b \!\,+\!\, d
\end{bmatrix}
\qquad\qquad
\begin{bmatrix}
1 \\ 3
\end{bmatrix}
+ \begin{bmatrix}
4 \\ 7
\end{bmatrix}
= \begin{bmatrix}
5 \\ 10
\end{bmatrix}
$$

Scalar Multiplication:
$$
k\!
\begin{bmatrix}
\phantom I \!\! a \\ b
\end{bmatrix}
=
\begin{bmatrix}
\phantom I \!\! ka \\ kb
\end{bmatrix}
\qquad\qquad
3\!
\begin{bmatrix}
\phantom I \!\! 2 \\ 3
\end{bmatrix}
=
\begin{bmatrix}
\phantom I \!\! 6 \\ 9
\end{bmatrix}
$$



### States as Vectors

Qubits are often represented with a column vector called a ***ket***.  

Each ket has an associated row vector called a ***bra***.

$$
\vert \psi \rangle = \left[ \begin{matrix}\phantom{A}\!\!\! \alpha \\ \beta\end{matrix} \right]
\qquad \text{and} \qquad
\langle \psi \vert = \left[ \begin{matrix} \,\alpha^* & \beta^* \, \end{matrix} \right]
$$

$\alpha$ and $\beta$ are complex numbers, and $\alpha^*$ and $\beta^*$ are their ***complex conjugates***.

Complex conjugation means that if $z = x + iy$, then $z^* = x - iy$.

Importantly, the state represented by $\alpha$ and $\beta$ is normalized:
$$
\vert \alpha \vert^2 + \vert \beta \vert ^2 = 1
$$

> Notes on terminology:  
> The term for converting a ket to a bra, or a bra to a ket, is taking the ***conjugate transpose***.    
> A quantum state follows the ***Euclidean*** norm.

### Vectors Need a Basis

We always choose a basis to work in.  The ***standard*** or ***computational basis*** states
in quantum computing are:

$$
\vert 0 \rangle = \begin{bmatrix} 1 \\ 0 \end{bmatrix}   \qquad
\text{and}
\qquad \vert 1 \rangle = \begin{bmatrix} 0 \\ 1 \end{bmatrix}
$$

A qubit in a ***pure state*** can be written in several equivalent ways:

$$
\vert \psi \rangle
\ \ =\ \
\alpha \vert 0 \!\: \rangle + \beta \vert 1\!\: \rangle
\ \ =\ \
\alpha \begin{bmatrix}1\\0\end{bmatrix} + \beta \begin{bmatrix}0\\1\end{bmatrix}
\ \ = \ \
\left[\begin{matrix}
\phantom{A}\!\!\!\alpha \\ \beta
\end{matrix}\right]
$$

This linear combination of basis states is called superposition.

> If a state can be described by a single state vector it is a pure state.
>
> The counterpart to pure states is the ***mixed state***, which will be introduced with noise on day five.


```python
a = 0.6 * Statevector.from_label('0')
b = 0.8 * Statevector.from_label('1')

psi = a + b

print(psi.is_valid())
```

### Inner Products

The inner product combines a row and a column vector to produce a scalar.
$$
\begin{bmatrix}
a & b
\end{bmatrix}
\begin{bmatrix}
c \\ d
\end{bmatrix}
= ac + bd
\qquad\qquad
\begin{bmatrix}
2 & 3
\end{bmatrix}
\begin{bmatrix}
4 \\ 1
\end{bmatrix}
= 11
$$


For quantum states the inner product is written using a bra and a ket, $\langle \psi \vert \psi \rangle$:
$$
\vert \psi \rangle = \left[ \begin{matrix}\phantom{A}\!\!\! \alpha \\ \beta\end{matrix} \right]
, \qquad
\quad
\langle \psi \vert = \left[ \begin{matrix} \,\alpha^* & \beta^* \, \end{matrix} \right]
$$

$$
\langle \psi \vert \psi \rangle
\ = \
\left[ \begin{matrix} \,\alpha^* & \beta^* \, \end{matrix} \right]
\left[ \begin{matrix}\phantom{A}\!\!\! \alpha \\ \beta\end{matrix} \right]
\ =\
\vert\alpha\vert^2 +
\vert \beta \vert^2 = 1
$$
$$
\langle 0 \vert 1 \rangle \ = \
\left[ \begin{matrix}1 & 0 \end{matrix} \right]
 \left[ \begin{matrix}0 \\ 1\end{matrix} \right]
=
0
$$
>
> Note that the equation for $\langle\psi\vert\psi\rangle$, demonstrates that the inner product of any *normalized* vector with itself is always 1.
>
> The inner product of two different orthonormal basis vectors with each other will always be zero.
>

```python
sv1 = Statevector.from_label('1') # |1>

print('The inner product <0|1>     = ', sv0.inner(sv1))
print('The inner product <psi|psi> = ', psi.inner(psi))
print('The inner product <psi|0>   = ', psi.inner(sv0))

# Note that Python uses "j" instead of "i" to indicate the imaginary unit, sqrt(-1).
```

### The Matrix

A matrix is a 2-dimensional array of numbers.


Matrix addition:
$$
\begin{bmatrix}
a & b \\
c & d
\end{bmatrix}
+
\begin{bmatrix}
e & f \\
g & h
\end{bmatrix}
=
\begin{bmatrix}
a\!\,+\!\,e & b\!\,+\!\,f \\
c\!\,+\!\,g & d\!\,+\!\,h
\end{bmatrix}
$$

Matrix-Vector product:
$$
\begin{bmatrix}
a & b \\
c & d
\end{bmatrix}
\begin{bmatrix}
x \\
y
\end{bmatrix}
=
\begin{bmatrix}
ax + by \\
cx + dy
\end{bmatrix}
$$

Matrix-Matrix product:
$$
\begin{bmatrix}
a & b \\
c & d
\end{bmatrix}
\begin{bmatrix}
e & f \\
g & h
\end{bmatrix}
=
\begin{bmatrix}
ae + bg & af + bh \\
ce + dg & cf + dh
\end{bmatrix}
$$


> Note that in both types of products the vector inner product is being used repeatedly to make each element in the result.

```python
# Working with matrices in Python is normally done with the NumPy package.
import numpy as np

I = np.array([[1, 0],
              [0, 1]])

X = np.array([[0, 1],
              [1, 0]])

print('I + X = \n', I+X)

print('\nI @ X =\n', I@X)  # The @ operator is used for matrix multiplication.
                             # Alternatively:   np.matmul(I, X)
```

### Matrices Act on States

In quantum computing operators are represented by matrices.  

They are linear transformations that act on the vectors representing quantum states:
$$
X = \begin{bmatrix}
0 & 1 \\
1 & 0
\end{bmatrix}
$$

If an operator $A$ acts on $\vert \psi \rangle$, then:  $\ \ A \vert \psi \rangle = \vert \psi' \rangle$
$$
X \vert 0 \rangle =
\begin{bmatrix}
0 & 1 \\
1 & 0
\end{bmatrix}
\begin{bmatrix}
1 \\ 0
\end{bmatrix} =
\begin{bmatrix}
0 \\ 1
\end{bmatrix}
$$

$$
X \vert 1 \rangle =
\begin{bmatrix}
0 & 1 \\
1 & 0
\end{bmatrix}
\begin{bmatrix}
0 \\ 1
\end{bmatrix} =
\begin{bmatrix}
1 \\ 0
\end{bmatrix}
$$

> This operator $X$ is known as the ***bit-flip*** operator.
>
> Not every matrix represents a physically allowed operation.  We will talk about the required properties soon.

```python
print('X|0> = |1> ? ', sv0.evolve(X) == sv1)
print('X|1> = |0> ? ', sv1.evolve(X) == sv0)
```

### Combining Operations

Multiple operators can be applied in sequence.
$$
\vert\psi\rangle
\ \xrightarrow{A}\  |\psi'\rangle
\ \xrightarrow{B}\  |\psi''\rangle
$$


So we have  $A \vert \psi \rangle = \vert \psi' \rangle$ and $B \vert \psi' \rangle = \vert \psi'' \rangle$.  

We can write this as:

$$
\vert \psi'' \rangle
\ =\
B\left(A\vert \psi \rangle\right)
\ =\
BA\vert \psi \rangle
$$

Since $BA$ is regular matrix multiplication we can calculate it first to simplify.

$$
BA = C, \qquad
\vert \psi'' \rangle
\ =\
BA\vert \psi \rangle
\ =\
C\vert \psi \rangle
$$


> Note how the operators are ordered.  The right-most operator is applied first.
>
> Matrices *ARE* associative, $A(BC) = (AB)C$, so we can muliply $BA$ before $A\vert\psi\rangle$ in the example above.
>
> Matrices *ARE NOT* generally commutative, $AB \ne BA$.  Changing their order can change the final state.

```python
XX = X.power(2)

print('XX = I ? ', XX == I)
```

### Two Types of Matrices

**Unitary operators** describe reversible evolutions of quantum states.  Matrices are unitary when:
$$
U^\dagger U = UU^\dagger = I
$$

***Hermitian operators*** describe ***observables***, their real eigenvalues are the possible measurement outcomes for that observable.
$$
A^\dagger = A
$$

> The symbol $\dagger$, read as “dagger,” indicates the conjugate transpose operation, also known as the adjoint.
>
> Multiplication by a unitary matrix preserves normalization when acting on a state vector.  Later you will see how this preserves probabilities.
>
> Hermitian operators are their own conjugate transpose. Because of this, their eigenvalues are real numbers.  Since those are the possible measurement outcomes of the operato, it is necessary they be real valued.

```python
print('X is unitary? ', X.is_unitary())      # X is unitary.
print('Alternatively:', X.dot(X.adjoint()) == I)
print()
print('X = X-dagger? ', X.adjoint() == X)    # X is also Hermitian.
```
